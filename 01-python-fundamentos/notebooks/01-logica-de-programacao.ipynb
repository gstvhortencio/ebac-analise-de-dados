```python
#usando f-string
valor = float(input("Quantas questões você acertou? "))
nota = min(valor*2, 10)
print(f"Então sua nota foi: {nota}")
#if
if nota > 7:
  print("Acima da média! Parabéns.")
#elif
elif nota == 7:
  print("Na média. Parabéns!")
#if-else
else:
  print("Vai ter que fazer recuperação. Boa sorte!")
```
