# Operadores Lógicos

*   And
*   Or
*   Not

# Tabela verdade

## And

```python
print(10 > 2 and 4 > 2)  # V and V
print(10 > 2 and 2 > 4)  # V and F
print(10 < 2 and 2 < 4)  # F and V
print(10 < 2 and 2 > 4)  # F and F
```

## Or

```python
print(10 > 2 or 4 > 2) # V and V
print(10 > 2 or 2 > 4) # V and F
print(10 < 2 or 2 < 4) # F and V
print(10 < 2 or 2 > 4) # F and F
```

## Not

```python
not 10 > 2 # V - F
```

```python
not 10 < 2 # F - V
```
