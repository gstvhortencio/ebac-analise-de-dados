#Cálculo de média

```python
##coleta das notas
nota1 = float(input("Digite a nota da primeira prova: "))
print("Ok.")
nota2 = float(input("Agora digite a nota da segunda prova: "))
print("Ok.")
nota3 = float(input("Por fim, digite a nota da terceira prova: "))
print("Aguarde sua média ser calculada.")

##soma para o cálculo da média
soma = nota1 + nota2 + nota3
##cálculo da média
media = soma / 3

##média utilizando f-string
print(f"A sua média é: {media}")
##mensagem final
if media >= 7:
  print("Parabéns! Você passou.")
else:
  print("Nos vemos no próximo semestre.")
```

#Verificação de senha simples

```python
##definir a senha permitida
senha_permitida = str("12345")
##solicitar senha do usuario
senha_usuario = str(input("Digite sua senha: "))
##verificar a permissão da senha
if senha_usuario == senha_permitida:
  print("Seja bem vindo!")
else:
  print("Senha incorreta. Tente novamente.")

##obs.: a senha é em string para que possamos utilizar todo tipo de caracter
```

#Lista

```python
##elaboração de listas
frutas = ["banana", "maçã", "pêra", "abacaxi", "uva", "morango", "melancia", "laranja"]

##exercício de feedback de clientes
##lista de possíveis feedbacks
feedbacks = ["Muito bom", "Bom", "Ruim", "Muito ruim"]

##dicionário com informações de um cliente
cliente1 = {"nome": "João", "idade": 30, "feedback": "Muito bom"}

##imprimir os feedbacks da lista
for feedback in feedbacks:
  print("- " + feedback)
##imprimir informações do cliente
print(f"\nnome: {cliente1['nome']}")
print(f"idade: {cliente1['idade']}")  # Remova int() se 'idade' já for um número
print(f"feedback: {cliente1['feedback']}")  # Remova str() se 'feedback' já for uma string
```

```python
##loop
for fruta in frutas:
  print(f"Eu gosto de {fruta}")
```
